# Regression Function

$$
  y = f(x_1, x_2, ..., x_n) + \epsilon
$$


- $f$- asılılıq funksiyası(x in y den asiliqindan sohbet gedir)
- $y$- hədəf (target) dəyişən(Bu datalar esasinda bu herdefe catmaga)
- $x_i$- xüsusiyyətlər (factor)
- $\epsilon$- səhv(küy)
misal ucun datalarimiz var funksiya bu datalarin esasinda qiymetini burda olan sapmalar kuyleri data olan sapmalari nezere alaraq bunu hesablayacaq


# Riyazi Forma

$$
  y = a + bx + \epsilon
$$

- $y$- hədəf (target) dəyişən(hansi hedefe gelib cixacayiq)
- $x$- asılı olmayan(müstəqil) dəyişən (factor, feature) (bu bizim elametlerimizdir misal ucun telebe  saat hazirlasibdi 5 saat hazirlasibdi bu datalarimizi)
- $a$- sərbəst dəyişən (intercept) (x 0 olanda bizim modelin proqnozlasdirdigi qiymetdi misal ucun derse qulaq asan 0 sifir faiz hazirlasibsa bu misalki 30 faiz cavab vere bilecey)
- $b$- əyilmə kooficienti (meyl əmsalı) (slope) (x 1 vahid artanda modelin proqnozunu nece vahid deyisdiyini bize gosterecek)
- $\epsilon$- təasdüfi səhv(küy) - real nəticə və proqnozlaşdırılmış qiymət arasında fərq
  - $y$ - real nəticə Məslən: 70
  - $\hat{y}$ - proqnoz nəticəsində alınan qiymət Məsələn: 68.7
  - bu zaman $\epsilon=1.3$

## 1. Scaler: MinMaxScaler və StandardScaler

Machine Learning-də istifadə etdiyimiz verilənlərin qiymətləri bir-birindən çox fərqli ola bilər. Məsələn, bir datasetdə yaş 18–60 arasında olduğu halda, maaş 500–8000 arasında ola bilər. Bu zaman bəzi alqoritmlər böyük qiymətlərə daha çox əhəmiyyət verə bilər.

Bu problemi azaltmaq üçün **Scaling** istifadə olunur. Scaling verilənlərin qiymətlərini müəyyən qaydada dəyişdirərək onları daha uyğun miqyasa gətirir. Ən çox istifadə olunan scaling üsullarından ikisi MinMaxScaler və StandardScaler-dir.

MinMaxScaler verilənləri əsasən 0 ilə 1 arasına gətirir. StandardScaler isə verilənləri elə dəyişir ki, onların orta qiyməti 0, standart sapması isə 1 olsun.

Məsələn, yaş və maaş kimi fərqli ölçüdə olan dəyişənləri modelə verməzdən əvvəl scaling etmək modelin bu dəyişənlərdən daha düzgün istifadə etməsinə kömək edə bilər.

### MinMaxScaler

MinMaxScaler verilənlərin qiymətlərini müəyyən intervala gətirmək üçün istifadə olunur. Ən çox istifadə olunan interval 0 ilə 1 arasındadır. Bu üsulda ən kiçik qiymət 0, ən böyük qiymət isə 1 olur. Digər qiymətlər isə bu iki qiymət arasında yerləşdirilir.

**Formula:**

$$
x_{scaled}=\frac{x-\min(x)}{\max(x)-\min(x)}
$$

Məsələn, `x = [10, 20, 30, 40, 50]` olduqda 30 qiyməti üçün:

$$
\frac{30-10}{50-10}=0.5
$$

Nəticədə:

- 10 → 0
- 20 → 0.25
- 30 → 0.50
- 40 → 0.75
- 50 → 1

**Kod nümunəsi:**

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
```

MinMaxScaler xüsusilə verilənlərin müəyyən intervalda olmasının vacib olduğu hallarda istifadə olunur. Kənar qiymətlər (outlier) olduqda digər qiymətlərin böyük hissəsi dar intervalda toplana bilər.

### StandardScaler

StandardScaler verilənləri standartlaşdırmaq üçün istifadə olunur. Bu üsulda verilənlərin orta qiymətinin 0, standart sapmasının isə 1 olması məqsəd qoyulur.

**Formula:**

$$
z=\frac{x-\text{mean}}{\text{standard deviation}}
$$

Məsələn, `x = [10, 20, 30, 40, 50]` üçün orta qiymət 30-dur. 30 qiyməti standartlaşdırıldıqda nəticə 0 olur.

**Kod nümunəsi:**

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

StandardScaler verilənləri orta qiymət və standart sapma əsasında standartlaşdırmaq lazım olduqda istifadə olunur. Bu üsul da kənar qiymətlərdən təsirlənə bilər.

### MinMaxScaler və StandardScaler arasındakı fərq

| Xüsusiyyət | MinMaxScaler | StandardScaler |
|---|---|---|
| Nəticə | Adətən 0–1 aralığı | Orta 0, standart sapma 1 |
| Əsas prinsip | Minimum və maksimum əsasında | Orta qiymət və standart sapma əsasında |
| Standart sapma | Müəyyən deyil | 1 olur |
| Outlier təsiri | Həssasdır | Həssasdır |
| Əsas məqsəd | Qiymətləri müəyyən intervala gətirmək | Qiymətləri standartlaşdırmaq |

**Sadə yadda saxlamaq üçün:** MinMaxScaler → bizim deyerleri mueyyen intervala yetirir 0-1 arasinda  mueyyen intervala yetirir ; StandardScaler → deyerleri orta qiymet etrafinda standartlasdirir

## 2. Factorize və One-Hot Encoding (OHE)

Machine Learning modelləri əsasən rəqəmlərlə işləyir. Datasetdə Red, Blue, Green kimi mətn tipli kateqoriyalar olduqda, onları modelin istifadə edə biləcəyi rəqəmsal formaya çevirmək lazımdır. Bu prosesə **Encoding** deyilir.

Encoding üçün müxtəlif üsullar var. Bunlardan ikisi Factorize və One-Hot Encoding (OHE)-dir.

### Factorize

Factorize kateqoriyalara rəqəmsal kod verir. Hər fərqli kateqoriyaya bir ədəd təyin olunur.

Məsələn:

- Red → 0
- Blue → 1
- Green → 2

**Pandas-da:**

```python
import pandas as pd

data = pd.Series(["Red", "Blue", "Green", "Red"])
encoded, categories = pd.factorize(data)
```

Factorize-in üstünlüyü odur ki, bir sütun saxlanılır və hər kateqoriyaya bir rəqəm verilir.

Lakin burada bir problem yarana bilər. Model 0, 1 və 2 rəqəmlərini müəyyən mənada sıralı və ya bir-birindən məsafəli qiymətlər kimi qəbul edə bilər.

Model bunu Blue-nun Red-dən, Green-in isə Blue-dan daha yüksək olduğu kimi qəbul edə bilər. Halbuki rənglər arasında belə bir sıra yoxdur.

### One-Hot Encoding (OHE)

One-Hot Encoding hər kateqoriya üçün ayrıca sütun yaradır. Hər sətirdə uyğun kateqoriyanın sütununa 1, digər sütunlara isə 0 yazılır.

Məsələn:

| Red | Blue | Green |
|---:|---:|---:|
| 1 | 0 | 0 |
| 0 | 1 | 0 |
| 0 | 0 | 1 |
| 1 | 0 | 0 |

**Pandas-da:**

```python
import pandas as pd

data = pd.DataFrame({
    "Color": ["Red", "Blue", "Green", "Red"]
})

result = pd.get_dummies(data, columns=["Color"])
```

Beləliklə, kateqoriyalar arasında süni sıralama yaranmır.

### Factorize və OHE arasındakı əsas fərq

| Xüsusiyyət | Factorize | One-Hot Encoding |
|---|---|---|
| Nəticə | Bir sütunda rəqəmlər | Bir neçə 0/1 sütunu |
| Sütun sayı | Dəyişmir | Arta bilər |
| Süni sıra yarada bilər? | Bəli | Xeyr |
| Sırasız kateqoriyalar | Bəzi modellərdə problem yarada bilər | Daha uyğundur |

### Hansı halda istifadə olunur?

Əgər kateqoriyalar arasında real sıralama yoxdursa, OHE daha uyğun seçimdir.

Məsələn:

Red, Blue, Green

**Nəticə:** Factorize kateqoriyaları rəqəmlərlə əvəz edir və bir sütun saxlanılır. Sadə və yaddaş baxımından rahatdır, lakin sırasız kateqoriyalarda süni sıra təsəvvürü yarada bilər. ## 4. Regression Metrics

Regression Metrics regression modelinin proqnozlarının nə qədər düzgün olduğunu ölçmək üçün istifadə olunan qiymətləndirmə göstəriciləridir.

Ən çox istifadə olunan regression metrikləri:

- MAE — Mean Absolute Error
- MSE — Mean Squared Error
- RMSE — Root Mean Squared Error
- R² Score — Coefficient of Determination

Bu metriklər eyni şeyi ölçmür. MAE, MSE və RMSE səhvin nə qədər olduğunu, R² isə modelin verilənlərdəki dəyişkənliyi nə qədər izah etdiyini göstərir.

### MAE — Mean Absolute Error

MAE modelin proqnozu ilə real qiymət arasındakı orta mütləq fərqi göstərir.

**Formula:**


$$
MAE = \frac{1}{n}\sum |y_i - \hat{y_i}|
$$


Məsələn, real qiymətlər 100, 200, 300, proqnozlar isə 90, 220, 280 olarsa, xətalar 10, 20, 20 olur.

$$
MAE=\frac{10+20+20}{3}=16.67
$$

Deməli, model orta hesabla 16.67 vahid səhv edir.

MAE-də mənfi və müsbət səhvlər bir-birini sıfırlamasın deyə mütləq qiymət istifadə olunur.

**Yaxşı nəticə:** 0-a yaxın MAE.

**Üstünlüyü:** Nəticəni başa düşmək asandır. Məsələn, MAE = 16.67-dirsə, modelin orta mütləq səhvinin təxminən 16.67 vahid olduğunu deyə bilərik.

### MSE — Mean Squared Error

MSE xətaların kvadratını götürür və onların orta qiymətini hesablayır.

**Formula:**

$$
MSE = \frac{1}{n}\sum (y_i - \hat{y_i})^2
$$

Yuxarıdakı nümunədə xətalar 10, 20 və 20-dir.

$$
MSE=\frac{10^2+20^2+20^2}{3}
$$

$$
MSE=\frac{100+400+400}{3}=300
$$

MSE böyük səhvləri daha ciddi cəzalandırır.

Məsələn, 5 vahidlik səhvin kvadratı 25, 20 vahidlik səhvin kvadratı isə 400 olur. Buna görə böyük səhvlər MSE nəticəsinə daha çox təsir edir.

**Yaxşı nəticə:** 0-a yaxın MSE.

### RMSE — Root Mean Squared Error

RMSE MSE-nin kvadrat köküdür.

**Formula:**

$$
RMSE=\sqrt{MSE}
$$

MSE 300 olduqda:

$$
RMSE=\sqrt{300}\approx17.32
$$

RMSE-nin əsas üstünlüyü nəticənin target dəyişəninin orijinal vahidində olmasıdır.

Məsələn, qiymət AZN ilə ölçülürsə, RMSE də AZN ilə ifadə olunur.

**MAE ilə fərqi:** RMSE böyük səhvləri daha çox cəzalandırır, MAE isə səhvlərin mütləq qiymətlərinin orta göstəricisini hesablayır.

**Yaxşı nəticə:** 0-a yaxın RMSE.

### R² Score — Coefficient of Determination

$$
R^2 = 1 - \frac{\text{modelin səhvi}}{\text{sadəcə orta qiymətdən istifadə edən modelin səhvi}}
$$

R² modelin target dəyişənindəki dəyişkənliyi nə qədər yaxşı izah etdiyini göstərir.

R² üçün nəticənin 1-ə yaxın olması adətən modelin yaxşı uyğunlaşdığını göstərir. R² = 0 olduqda model test məlumatlarında sadəcə real qiymətlərin orta qiymətindən istifadə edən yanaşma ilə eyni nəticə verə bilər. R² mənfi də ola bilər və bu, modelin həmin müqayisə üsulundan daha pis nəticə verdiyini göstərir.

Məsələn, R² = 0.90 olduqda model target dəyişənindəki dəyişkənliyin təxminən 90%-ni izah edir.

**Yaxşı nəticə:** 1-ə yaxın R².

### MAE, MSE, RMSE və R² fərqi

| Metrik | Nə ölçür? | Əsas xüsusiyyət | Yaxşı nəticə |
|---|---|---|---|
| MAE | Orta mütləq səhv | Səhvlərin orta mütləq ölçüsünü göstərir | 0-a yaxın |
| MSE | Kvadratlaşdırılmış orta səhv | Böyük səhvləri daha güclü cəzalandırır | 0-a yaxın |
| RMSE | MSE-nin kvadrat kökü | Orijinal ölçü vahidində nəticə verir və böyük səhvlərə həssasdır | 0-a yaxın |
| R² | Modelin izah etdiyi dəyişkənlik | Modelin uyğunluğunu göstərir | 1-ə yaxın |

**Yadda saxlamaq üçün:**

MAE — orta mütləq səhv.

MSE — xətaların kvadratlarının ortası.

RMSE — MSE-nin kvadrat kökü.

R² — modelin verilənlərdəki dəyişkənliyi nə qədər izah etdiyini göstərir.

Ümumilikdə, regression modelini qiymətləndirərkən tək bir metrikə baxmaq əvəzinə bir neçə metrikdən birlikdə istifadə etmək daha məqsədəuyğundur.
